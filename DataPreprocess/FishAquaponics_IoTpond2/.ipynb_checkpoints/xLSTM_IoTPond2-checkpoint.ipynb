```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import matplotlib.pyplot as plt

# 讀取資料
df = pd.read_csv('cleaned_IoTPond2.csv')

# 特徵與目標欄位
features = ['temperature', 'turbidity', 'disolved_oxg', 'ph', 'ammonia', 'nitrate', 'fish_length']
target = 'fish_weight'

# 標準化
scaler_x = MinMaxScaler()
scaler_y = MinMaxScaler()
X_scaled = scaler_x.fit_transform(df[features])
y_scaled = scaler_y.fit_transform(df[[target]])

# 建立序列資料
sequence_length = 60
X_seq, y_seq = [], []
for i in range(len(X_scaled) - sequence_length):
    X_seq.append(X_scaled[i:i + sequence_length])
    y_seq.append(y_scaled[i + sequence_length])
X_seq = np.array(X_seq)
y_seq = np.array(y_seq)

# 分訓練與測試集
split = int(len(X_seq) * 0.8)
X_train, X_test = X_seq[:split], X_seq[split:]
y_train, y_test = y_seq[:split], y_seq[split:]

# 轉換為 tensor
X_train_tensor = torch.tensor(X_train, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train, dtype=torch.float32)
X_test_tensor = torch.tensor(X_test, dtype=torch.float32)
y_test_tensor = torch.tensor(y_test, dtype=torch.float32)

# 建立 DataLoader
batch_size = 64
train_loader = DataLoader(TensorDataset(X_train_tensor, y_train_tensor), batch_size=batch_size, shuffle=True)

# 定義 xLSTM 模型（簡化版）
class xLSTM(nn.Module):
    def __init__(self, input_size, hidden_size, num_layers, output_size):
        super(xLSTM, self).__init__()
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers, batch_first=True)
        self.fc = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        out, _ = self.lstm(x)
        out = out[:, -1, :]
        out = self.fc(out)
        return out

# 模型參數
input_size = len(features)
hidden_size = 64
num_layers = 2
output_size = 1
model = xLSTM(input_size, hidden_size, num_layers, output_size)

# 損失函數與優化器
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

# 訓練模型
epochs = 50
for epoch in range(epochs):
    model.train()
    epoch_loss = 0
    for X_batch, y_batch in train_loader:
        optimizer.zero_grad()
        outputs = model(X_batch)
        loss = criterion(outputs, y_batch)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()
    print(f'Epoch {epoch+1}/{epochs}, Loss: {epoch_loss/len(train_loader):.6f}')

# 測試模型
model.eval()
with torch.no_grad():
    y_pred = model(X_test_tensor).numpy()
    y_pred_rescaled = scaler_y.inverse_transform(y_pred)
    y_test_rescaled = scaler_y.inverse_transform(y_test)

# 計算指標
r2 = r2_score(y_test_rescaled, y_pred_rescaled)
mse = mean_squared_error(y_test_rescaled, y_pred_rescaled)
mae = mean_absolute_error(y_test_rescaled, y_pred_rescaled)

print(f'R² Score: {r2:.4f}')
print(f'MSE: {mse:.4f}')
print(f'MAE: {mae:.4f}')

# 繪圖比較
plt.figure(figsize=(10, 6))
plt.plot(y_test_rescaled, label='True Fish Weight')
plt.plot(y_pred_rescaled, label='Predicted Fish Weight')
plt.legend()
plt.xlabel('Sample')
plt.ylabel('Fish Weight')
plt.title(f'Fish Weight Prediction (xLSTM)\nR²: {r2:.4f}, MSE: {mse:.4f}, MAE: {mae:.4f}')
plt.show()
```

## Please use the hint above to finish the following questions  
### Q1-1: Create a linear regression model to predict the total fare.
### Q1-2: Evaluate the predictive result of this model by R<sup>2</sup>, MSE, and MAE metrics.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.model_selection import train_test_split

# 讀取資料
df = pd.read_csv('https://github.com/QaidJamali53/Railways-Price-Prediction/raw/refs/heads/main/price_data.csv.zip',compression='zip')

# 移除不適用的欄位
df_clean = df.drop(columns=['availability', 'timeStamp', 'fromStnCode', 'toStnCode'])

# 將類別變數進行標籤編碼
df_clean['classCode'] = LabelEncoder().fit_transform(df_clean['classCode'])

# 資料清理-處理缺失值與無效值
df_clean.replace([np.nan, np.inf, -np.inf], 0, inplace=True)

# 切分特徵與標籤
X = df_clean.drop(columns=['totalFare'])
y = df_clean['totalFare']

# 資料標準化 (Z-score)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 訓練與測試資料切分
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# 建立線性迴歸模型並訓練
model = LinearRegression()
model.fit(X_train, y_train)

# 預測與評估
y_pred = model.predict(X_test)

r2 = r2_score(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)

# 輸出評估指標
print(f"\033[94m Linear Regression Model Evaluation Metrics: \033[0m")
print(f"\033[94m R² Score: {r2:.6f} \033[0m")
print(f"\033[94m Mean Squared Error (MSE): {mse:.4f} \033[0m")
print(f"\033[94m Mean Absolute Error (MAE): {mae:.4f} \033[0m")
```

### Q2-1: Create a LASSO regression model to predict the total fare.
### Q2-2: Evaluate the predictive result of this model by R<sup>2</sup>, MSE, and MAE metrics.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import Lasso
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.model_selection import train_test_split

# 讀取資料
df = pd.read_csv('https://github.com/QaidJamali53/Railways-Price-Prediction/raw/refs/heads/main/price_data.csv.zip',compression='zip')

# 移除不適用的欄位
df_clean = df.drop(columns=['availability', 'timeStamp', 'fromStnCode', 'toStnCode'])

# 類別變數進行標籤編碼
df_clean['classCode'] = LabelEncoder().fit_transform(df_clean['classCode'])

# 處理缺失與無效值
df_clean.replace([np.nan, np.inf, -np.inf], 0, inplace=True)

# 切分特徵與標籤
X = df_clean.drop(columns=['totalFare'])
y = df_clean['totalFare']

# 資料標準化 (Z-score)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 切分訓練與測試資料
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# 建立LASSO迴歸模型並訓練（具備L1正則化的線性模型）
model = Lasso(alpha=0.1)  # alpha越大，正則化越強
model.fit(X_train, y_train)

# 預測與評估
y_pred = model.predict(X_test)

r2 = r2_score(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)

# 顯示評估指標
print(f"\033[94m LASSO Regression Model Evaluation Metrics: \033[0m")
print(f"\033[94m R² Score: {r2:.6f} \033[0m")
print(f"\033[94m Mean Squared Error (MSE): {mse:.4f} \033[0m")
print(f"\033[94m Mean Absolute Error (MAE): {mae:.4f} \033[0m")
```

### Q3-1: Create a kNN regression model to predict the total fare.
### Q3-2: Evaluate the predictive result of this model by R<sup>2</sup>, MSE, and MAE metrics.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.model_selection import train_test_split

# 讀取資料
df = pd.read_csv('https://github.com/QaidJamali53/Railways-Price-Prediction/raw/refs/heads/main/price_data.csv.zip',compression='zip')

# 移除不適用的欄位
df_clean = df.drop(columns=['availability', 'timeStamp', 'fromStnCode', 'toStnCode'])

# 類別變數進行標籤編碼
df_clean['classCode'] = LabelEncoder().fit_transform(df_clean['classCode'])

# 處理缺失與無效值
df_clean.replace([np.nan, np.inf, -np.inf], 0, inplace=True)

# 切分特徵與標籤
X = df_clean.drop(columns=['totalFare'])
y = df_clean['totalFare']

# 資料標準化 (Z-score)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 切分訓練與測試資料
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# 建立KNN迴歸模型並訓練
model = KNeighborsRegressor(n_neighbors=3)
model.fit(X_train, y_train)

# 預測與評估
y_pred = model.predict(X_test)

r2 = r2_score(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)

# 顯示評估指標
print(f"\033[94m KNN Regression Model Evaluation Metrics: \033[0m")
print(f"\033[94m R² Score: {r2:.6f} \033[0m")
print(f"\033[94m Mean Squared Error (MSE): {mse:.4f} \033[0m")
print(f"\033[94m Mean Absolute Error (MAE): {mae:.4f} \033[0m")
```

```python
import pandas as pd
import numpy as np
```

```python
# 載入三個檔案
df_normalized = pd.read_csv('cleaned_IoTPond2_normalized.csv')
df_features = pd.read_csv('scaled_feature_data.csv')
df_targets = pd.read_csv('scaled_target_data.csv')
# df_normalized
```

```python
# 檢查維度
shape_normalized = df_normalized.shape
shape_features = df_features.shape
shape_targets = df_targets.shape

# 檢查形狀
print('Normalized shape:', df_normalized.shape)
print('Feature shape:', df_features.shape)
print('Target shape:', df_targets.shape)
```

```python
# 檢查列數是否一致
if df_normalized.shape[0] == df_features.shape[0] == df_targets.shape[0]:
    print('✅ 列數一致')
else:
    print('❌ 列數不一致')
```

```python
# 檢查欄位總數是否對應
if df_features.shape[1] + df_targets.shape[1] == df_normalized.shape[1]:
    print('✅ 欄位總數一致')
else:
    print('❌ 欄位總數不一致')
```
